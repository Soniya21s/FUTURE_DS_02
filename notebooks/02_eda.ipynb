```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/processed/cleaned_data.csv")
```

```python
df.head(2)
```

## TARGET ANALYSIS

```python
sns.countplot(x='Churn', data=df)
plt.title("Churn Distribution")
plt.show()

print(df['Churn'].value_counts(normalize=True)*100)
```

### KEY FEATURE vs CHURN

1. Previously Insured

```python
pd.crosstab(df['Previously_Insured'], df['Churn'], normalize='index') * 100
```

2. Vehicle Damage

```python
pd.crosstab(df['Vehicle_Damage'], df['Churn'], normalize='index') * 100
```

3. Vehicle Age

```python
pd.crosstab(df['Vehicle_Age'], df['Churn'], normalize='index') * 100
```

4. Gender

```python
pd.crosstab(df['Gender'], df['Churn'], normalize='index') * 100
```

5. Age Buckets

```python
df['Age_Group'] = pd.cut(df['Age'], bins=[20,30,40,50,60,70,90])

pd.crosstab(df['Age_Group'], df['Churn'], normalize='index') * 100
```

6. Vintage

```python
df['Vintage_Group'] = pd.cut(df['Vintage'], bins=[0,100,200,300])

pd.crosstab(df['Vintage_Group'], df['Churn'], normalize='index') * 100
```
